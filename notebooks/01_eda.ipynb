**Columns of the dataset and what they mean:**


- `date` &#8594; The date of when the record was taken

- `adm_level` &#8594; The adminstrative level of the data, 1 is provincial level and 2 is district level
- `adm_id` &#8594; The numeric ID for the administrative unit
- `PCODE` &#8594; The Place code used by OCHA/UN; is the standard identifier for the dataset
- `n_pixels` &#8594; The number of satellite raster pixels that fall inside the district/province.
- `rfh` &#8594; 10-day rainfall [mm]
- `r1h` &#8594; Rainfall 1-month rolling aggregation [mm]
- `r3h` &#8594; Rainfall 3-month rolling aggregation [mm]
- `rfh_avg` &#8594; Long-term average of 10-day rainfall [mm]
- `r1h_avg` &#8594; Long-term average of 1-month rolling rainfall [mm]
- `r3h_avg` &#8594; Long-term average of 3-month rolling rainfall [mm]
- `rfq` &#8594; 10-day rainfall anomaly [%]
- `r1q` &#8594; 1-month rainfall anomaly [%]
- `r3q` &#8594; 3-month rainfall anomaly [%]
