**Columns of the dataset and what they mean:**

- `date` &#8594; Dekad start date (01, 11, 21 of each month)
- `adm_level` &#8594; 1 = province, 2 = district
- `adm_id` &#8594; Numeric ID for the administrative unit
- `PCODE` &#8594; OCHA/UN place code; standard identifier for the dataset
- `n_pixels` &#8594; Number of satellite raster pixels inside the district/province
- `rfh` &#8594; 10-day rainfall as a **ratio to the unit's long-term mean** [%]
- `r1h` &#8594; **1-month rolling** rainfall [mm]
- `r3h` &#8594; **3-month rolling** rainfall [mm]
- `rfh_avg` &#8594; Long-term average of 10-day rainfall (climatology) [mm]
- `r1h_avg` &#8594; Long-term average of 1-month rolling rainfall [mm]
- `r3h_avg` &#8594; Long-term average of 3-month rolling rainfall [mm]
- `rfq` &#8594; 10-day rainfall anomaly [%]
- `r1q` &#8594; 1-month rainfall anomaly [%]
- `r3q` &#8594; 3-month rainfall anomaly [%]
- `version` &#8594; `final` or `prelim` (provisional tail)

## EDA

Structure checks, missing-value handling, and five charts.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style='whitegrid', context='notebook')
pd.set_option('display.max_columns', 30)

# Re-read the file WITH parse_dates. The first two cells loaded it without it,
# which leaves `date` as text: string comparison ('1981-04-01' >= '1981-03-21')
# happens to look right for ISO dates, but `df.date.dt.month` and
# `df.date.min().date()` both fail outright. Parsing once here means every
# later cell can treat the column as a real datetime.
df = pd.read_csv('../data/nepal_rainfall_dataset.csv', parse_dates=['date'])
print(df.shape)
```

### 1. Panel structure

Is this a clean, balanced panel? One row per unit per dekad, no gaps, no duplicates?

```python
n_dupes = df.duplicated(['adm_id', 'date']).sum()
per_unit = df.groupby('adm_id').size()

print(f"Rows                     : {len(df):,}")
print(f"Date range               : {df.date.min().date()} -> {df.date.max().date()}")
print(f"Distinct dekad dates     : {df.date.nunique():,}")
print(f"Admin units              : {df.adm_id.nunique()}  "
      f"({df[df.adm_level == 1].adm_id.nunique()} provinces, "
      f"{df[df.adm_level == 2].adm_id.nunique()} districts)")
print(f"Duplicate (unit, date)   : {n_dupes}")
print(f"Observations per unit    : min={per_unit.min()}, max={per_unit.max()}")
print(f"n_pixels constant/unit   : {(df.groupby('adm_id').n_pixels.nunique() == 1).all()}")
print()
print('Rows per version:\n', df.version.value_counts().to_string())
```

```python
miss = df.isna().sum()
print('Missing values:\n', miss[miss > 0].to_string())

# Which dates are affected, and how many units on each date?
for col in ['r1h', 'r3h']:
    bad = df.loc[df[col].isna(), 'date']
    print(f"\n{col}: {bad.nunique()} distinct dates, {bad.max().date()} is the last "
          f"affected date -> a {bad.nunique()}-dekad warm-up window")

# r1h needs 3 dekads, r3h needs 9; both start at the very beginning of the record.
print('\nWarm-up window is at the START of the series, not scattered through it,')
print('so these are undefined-by-construction cells and must NOT be imputed.')
```

```python
# HANDLING THE MISSING VALUES
#
# 1. r1h / r3h warm-up  -> drop the first 9 dekads (1981-01-01 .. 1981-03-21).
#    Every column is then well defined. Dropping is correct because the value is
#    genuinely undefined; imputing it would invent a month of rainfall.
# 2. version == 'prelim' -> the last two dekads are provisional and revisable by
#    the publisher. Excluded so results are reproducible.
# 3. rfh == 0            -> 82 rows in 5 units, always on 11-11 / 11-21. Zero rain
#    in 45 consecutive Novembers is a "no retrieval" sentinel, not weather. Flag it.
# 4. adm_level           -> provinces and districts are PARALLEL aggregations, not
#    nested totals, so they are never pooled into one feature matrix.

SENTINEL_UNITS = {'NP0104', 'NP0325', 'NP0546', 'NP0547', 'NP0548'}

df['is_sentinel_zero'] = (df.rfh == 0) & df.PCODE.isin(SENTINEL_UNITS)
n_sentinel = int(df.is_sentinel_zero.sum())
print(f"Sentinel-zero rows: {n_sentinel} in {df.loc[df.is_sentinel_zero, 'PCODE'].nunique()} units, "
      f"on {sorted(df.loc[df.is_sentinel_zero, 'date'].dt.strftime('%m-%d').unique())}")

n_start = len(df)
df = (df[df.version == 'final']
        .loc[lambda d: d.date >= '1981-04-01']
        .sort_values(['adm_id', 'date'])
        .reset_index(drop=True))

print(f"\nRows: {n_start:,} -> {len(df):,}")
print(f"Remaining NaNs: {int(df.isna().sum().sum())}")
print(f"Date range now : {df.date.min().date()} -> {df.date.max().date()}")

districts = df[df.adm_level == 2].copy()
print(f"District subset: {len(districts):,} rows, {districts.adm_id.nunique()} districts")
```

### 3. Distributions

`rfh` is a ratio with a long tail. The log1p transform below is used later for
modelling, but all descriptive statistics here are on the original scale.

```python
targets = ['rfh', 'r1h', 'r3h']
print(districts[targets].describe().T.round(2).to_string())
print()
print('Skewness / kurtosis:')
for t in targets:
    s = districts[t]
    print(f"  {t:5s} skew={s.skew():6.2f}  kurtosis={s.kurtosis():7.2f}  "
          f"IQR fence upper={s.quantile(.75) + 1.5 * (s.quantile(.75) - s.quantile(.25)):8.1f}  "
          f"share above={(s > s.quantile(.75) + 1.5 * (s.quantile(.75) - s.quantile(.25))).mean():.2%}")
```

## Charts

### Chart 1 — Distribution of `rfh`

Heavy right tail: the monsoon tail is the flood signal and must not be clipped away.

```python
fig, axes = plt.subplots(1, 2, figsize=(14, 4.5))

sns.histplot(districts.rfh, bins=80, kde=True, ax=axes[0], color='#2E6F9E')
axes[0].set_yscale('log')
axes[0].set_title('rfh is strongly right-skewed (log y-axis)')
axes[0].set_xlabel('rfh  (ratio to long-term mean, %)')

sns.histplot(np.log1p(districts.rfh), bins=80, kde=True, ax=axes[1], color='#C1443B')
axes[1].set_title('log1p(rfh) is close to symmetric')
axes[1].set_xlabel('log1p(rfh)')

plt.tight_layout()
plt.show()
```

### Chart 2 — Seasonal cycle

Monthly mean `rfh`. This is the single dominant pattern in the dataset, and it is
why a seasonal climatology is such a strong baseline.

```python
d = districts.assign(month=districts.date.dt.month)
season = d.groupby('month').rfh.agg(['mean', 'std'])
season['ci'] = season['std'] / np.sqrt(len(d) / 12)

fig, ax = plt.subplots(figsize=(12, 4.5))
ax.bar(season.index, season['mean'], yerr=season['ci'], capsize=3,
       color='#3B7EA1', edgecolor='white')
ax.set_xticks(season.index)
ax.set_xticklabels(['Jan','Feb','Mar','Apr','May','Jun','Jul','Aug','Sep','Oct','Nov','Dec'])
ax.set_ylabel('mean rfh')
ax.set_title('Monsoon dominance: July mean is ~80x the November mean')
for m, v in season['mean'].items():
    ax.text(m, v + 4, f'{v:.1f}', ha='center', fontsize=9)
plt.tight_layout()
plt.show()

print(season['mean'].round(2).to_string())
```

### Chart 3 — Trend over time

Annual mean `rfh`. With no significant trend, the target has no level to learn and
any predictive power must come from seasonality.

```python
annual = districts.assign(year=districts.date.dt.year).groupby('year').rfh.mean()
# 2026 is a partial year (data end 2026-08-21) and prelim rows are already
# dropped, so exclude it from the trend fit rather than letting it bias the slope.
full = annual.drop(index=annual.index.max())

fig, ax = plt.subplots(figsize=(13, 4.5))
ax.plot(annual.index, annual.values, marker='o', ms=4, color='#2E6F9E', lw=1.4)
ax.plot(full.index, np.poly1d(np.polyfit(full.index, full.values, 1))(full.index),
        color='#C1443B', lw=1.8, label='linear fit 1981-2025')
ax.axhline(full.mean(), ls=':', color='grey', lw=1.2, label=f'mean {full.mean():.1f}')
ax.set_xlabel('year')
ax.set_ylabel('mean rfh')
ax.set_title('No meaningful long-term trend in rfh')
ax.legend(frameon=False)
plt.tight_layout()
plt.show()

slope = np.polyfit(full.index, full.values, 1)[0]
print(f"Linear slope over {full.index.min()}-{full.index.max()}: {slope:+.4f} rfh per year")
print(f"National mean rfh has not moved: the whole signal is seasonal, not a trend.")

# Does last year tell you anything about this year? (within-unit, year over year)
unit_year = districts.assign(year=districts.date.dt.year).groupby(['adm_id', 'year']).rfh.mean().unstack()
pairs = pd.concat(
    [pd.concat([unit_year[y], unit_year[y + 1]], axis=1, keys=['this', 'next'])
     for y in unit_year.columns[:-1] if (y + 1) in unit_year.columns]
).dropna()
print(f"Year-over-year correlation of district mean rfh: {pairs['this'].corr(pairs['next']):.3f} "
      f"(n = {len(pairs):,} district-years)")
```

### Chart 4 — Correlation between rainfall columns

The numbers to read carefully: `rfq` vs `rfh` is **1.00** because `rfq` is
`100 × rfh / rfh_avg` — an algebraic inverse of the target, not an independent
measurement. Any model given `rfq` and `rfh_avg` scores R² ≈ 0.998 with no
learning at all.

```python
cols = ['rfh', 'rfh_avg', 'r1h', 'r1h_avg', 'r3h', 'r3h_avg', 'rfq', 'r1q', 'r3q']
corr = districts[cols].corr()

# Correlation with unit and season removed — the honest test for derivation.
districts['cell'] = districts.PCODE + '|' + districts.date.dt.strftime('%m-%d')
within = districts.groupby('cell')[cols].corr().groupby(level=1).mean()

# A climatology is CONSTANT inside each (unit x month-day) cell, so its
# within-cell standard deviation is zero and Pearson correlation is undefined ->
# NaN. NaN here means "carries literally no within-cell information", which is
# the strongest possible statement, not a computation failure. Record it as 0.
n_const = within.isna().sum().sum()
within = within.fillna(0.0)
print(f"{int(n_const)} column-cells are exactly constant within their cell "
      f"(the three *_avg climatologies) -> recorded as 0.0000")

fig, axes = plt.subplots(1, 2, figsize=(17, 6.5))
for ax, mat, title in [
        (axes[0], corr, f'Raw correlation{chr(10)}(dominated by the shared seasonal cycle)'),
        (axes[1], within, f'Correlation WITHIN (unit x month-day){chr(10)}(true derivation)')]:
    sns.heatmap(mat, annot=True, fmt='.2f', cmap='RdBu_r', center=0,
                vmin=-1, vmax=1, ax=ax, square=True, linewidths=.4, cbar_kws={'shrink': .8})
    ax.set_title(title, fontsize=11)

plt.tight_layout()
plt.show()

print('Leakage verdict (within-cell correlation with rfh):')
print(within['rfh'].drop('rfh').sort_values(ascending=False).round(4).to_string())
```

### Chart 5 — Spatial and seasonal structure

District × month mean `rfh`. This is the pattern a seasonal climatology already
captures, which is why the baselines are so hard to beat.

```python
d = districts.assign(month=districts.date.dt.month)

# Rows = districts, columns = months, value = mean rfh. This is the single
# picture that shows both dimensions of the problem at once.
piv = d.pivot_table(index='PCODE', columns='month', values='rfh', aggfunc='mean')

# Sort districts by their July value: the monsoon peak, i.e. the clearest
# single number for "how wet is this place".
row_order = piv[7].sort_values(ascending=False).index
piv = piv.loc[row_order]

fig, axes = plt.subplots(1, 2, figsize=(17, 11),
                         gridspec_kw={'width_ratios': [2.4, 1]})

sns.heatmap(piv, ax=axes[0], cmap='YlGnBu', linewidths=.3,
            cbar_kws={'label': 'mean rfh', 'shrink': .6})
axes[0].set_title('District x month mean rfh, wettest-July district at the top', fontsize=12)
axes[0].set_xlabel('month')
axes[0].set_ylabel('district (PCODE)')

# Right panel: is a wet district also a more erratic one?
spread = d.groupby('PCODE').rfh.agg(['mean', 'std']).sort_values('mean')
axes[1].scatter(spread['mean'], spread['std'], s=26, alpha=.75, color='#C1443B')
axes[1].set_xlabel('district mean rfh')
axes[1].set_ylabel('district std rfh')
axes[1].set_title('Wet districts are also\nthe more variable ones', fontsize=11)
for code_, r in spread.iterrows():
    axes[1].annotate(code_, (r['mean'], r['std']), fontsize=5, alpha=.7,
                     xytext=(2, 2), textcoords='offset points')
axes[1].set_xlim(left=0)
axes[1].set_ylim(bottom=0)

plt.tight_layout()
plt.show()

print(f"District mean rfh: min={spread['mean'].min():.1f} ({spread['mean'].idxmin()}), "
      f"max={spread['mean'].max():.1f} ({spread['mean'].idxmax()})")
print(f"corr(district mean rfh, district std rfh) = {spread['mean'].corr(spread['std']):.3f}")
```

## Key findings

1. **The panel is clean and perfectly balanced.** 138,264 rows = 84 units x 1,646 dekads
   (1981-01-01 to 2026-09-11), zero duplicate `(unit, date)` pairs, and `n_pixels` constant
   within each unit. There is no structural tidying to do.

2. **Every missing value is structural, not random.** `r1h` / `r1h_avg` / `r1q` are
   undefined for the first 2 dekads and `r3h` / `r3h_avg` / `r3q` for the first 8, because
   they are 3- and 9-dekad trailing sums that need history before they exist. The gaps sit
   only at the very start of the record, so dropping the first 9 dekads leaves **zero** NaNs
   and invents nothing. Imputing them would fabricate a month of rainfall.

3. **168 provisional rows were excluded** (`version == 'prelim'`, 2026-09-01 and 2026-09-11).
   The publisher can revise them, so any result depending on them is not reproducible.

4. **82 values are "no retrieval", not weather.** `rfh == 0` for 5 units on 11-11 and 11-21
   in every single year. Flagged as `is_sentinel_zero` and excluded from statistics; treating
   them as real drought would drag those districts' November means toward zero forever.

5. **`rfh` is a ratio, not a depth, and it is heavily right-skewed** (skew 1.91, kurtosis
   4.40). 5.25% of values sit above the IQR fence - that tail is the flood signal, so it is
   kept and later modelled on a `log1p` scale rather than clipped.

6. **Seasonality is the dominant signal and there is no trend.** July mean 134.1 against
   November 1.7; the national annual mean moves by -0.02 rfh per year across 45 years. So a
   model cannot learn a level or a direction - only *when* a district is wet relative to its
   own normal. Year-over-year district means still correlate 0.736, so recent years carry
   real information.

7. **Target leakage is present in the file.** Holding unit and season fixed, `rfq`
   correlates **1.0000** with `rfh` because `rfq` is `100 x rfh / rfh_avg`, an algebraic
   inverse of the target - a model given `rfq` and `rfh_avg` recovers `rfh` at R2 = 0.998
   without learning anything. `r1h` and `r3h` correlate 0.645 and 0.376 because their
   trailing windows *contain* the current dekad, so they are usable only when lagged. The
   three `*_avg` climatologies score exactly 0.0000: they are constant inside each
   `(unit, month-day)` cell and carry no year-to-year information at all.

8. **Provinces and districts are parallel, not nested.** Province `rfh` is only ~9% of the
   sum of its districts' `rfh`, while `n_pixels` is exactly additive. The two levels
   therefore cannot be pooled into one feature matrix, and every model must fix one level.

9. **Wet districts are also the most volatile** (district mean vs std `rfh`, r = 0.948), and
   district means span 20.1 (NP0666) to 64.0 (NP0104). This single gradient explains most of
   the structure a clustering algorithm will find in Section 5.
